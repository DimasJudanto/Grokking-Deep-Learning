## 1. Neural Network Paling Sederhana

Bentuk neural network paling sederhana terdiri dari satu input, satu weight, dan satu prediction.

Rumusnya:

$$prediction = input \times weight$$

Contoh:

- input = `8.5`
- weight = `0.1`

Maka:

$$prediction = 8.5 \times 0.1 = 0.85$$

Pada tahap ini weight sudah tersedia. Network hanya melakukan perhitungan untuk menghasilkan prediction.

````python
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5, 9.5, 10, 9]

input = number_of_toes[0]
pred = neural_network(input, weight)

print(pred)
````

### Hasil

Output:

```text
0.85
```

Prediction diperoleh dengan mengalikan input `8.5` dengan weight `0.1`.

Weight dapat dipahami sebagai parameter yang menentukan seberapa besar pengaruh input terhadap prediction.

## 2. Multiple Inputs

Satu prediction dapat dibuat dari beberapa input. Setiap input mempunyai weight masing-masing.

Contoh:

```text
inputs  = [# toes, % win, # fans]
weights = [weight1, weight2, weight3]
```

Prediction dihitung menggunakan **weighted sum**:

$$prediction = \sum_i input_i \times weight_i$$

Dengan contoh:

$$
(8.5)(0.1)+(0.65)(0.2)+(1.2)(0.0)
=0.98
$$

Jadi setiap input memberikan kontribusi terhadap prediction sesuai dengan weight-nya.

````python
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0
    for i in range(len(a)):
        output += a[i] * b[i]
    return output

weights = [0.1, 0.2, 0.0]

def neural_network(input, weights):
    pred = w_sum(input, weights)
    return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0], wlrec[0], nfans[0]]
pred = neural_network(input, weights)

print(pred)
````

### Hasil

Output:

```text
0.98
```

Perhitungannya:

- `8.5 × 0.1 = 0.85`
- `0.65 × 0.2 = 0.13`
- `1.2 × 0.0 = 0`

Sehingga:

$$0.85+0.13+0=0.98$$

Operasi inilah yang menjadi dasar **weighted sum** pada neural network.

## 3. Dot Product dengan NumPy

Weighted sum dapat dihitung menggunakan **dot product**. Dot product melakukan perkalian elemen yang bersesuaian kemudian menjumlahkan hasilnya.

Secara matematis:

$$a \cdot b = \sum_i a_i b_i$$

Operasi ini akan sering digunakan ketika neural network mulai menggunakan vector dan matrix.

````python
import numpy as np

weights = np.array([0.1, 0.2, 0.0])

def neural_network(input, weights):
    pred = input.dot(weights)
    return pred

toes = np.array([8.5, 9.5, 9.9, 9.0])
wlrec = np.array([0.65, 0.8, 0.8, 0.9])
nfans = np.array([1.2, 1.3, 0.5, 1.0])

input = np.array([toes[0], wlrec[0], nfans[0]])
pred = neural_network(input, weights)

print(pred)
````

### Hasil

Output tetap:

```text
0.98
```

NumPy hanya membuat perhitungan weighted sum menjadi lebih ringkas melalui operasi vector/dot product.

## 4. Multiple Outputs

Satu input juga dapat menghasilkan beberapa prediction.

Misalnya satu input `% win` digunakan untuk menghasilkan tiga output:

- `hurt`
- `win`
- `sad`

Setiap output mempunyai weight sendiri.

Untuk satu input dan tiga output, operasi dapat ditulis sebagai **elementwise multiplication**:

$$prediction_i = input \times weight_i$$

````python
def ele_mul(number, vector):
    output = [0, 0, 0]
    assert(len(output) == len(vector))

    for i in range(len(vector)):
        output[i] = number * vector[i]

    return output

def neural_network(input, weights):
    pred = ele_mul(input, weights)
    return pred

wlrec = [0.65, 0.8, 0.8, 0.9]
weights = [0.3, 0.2, 0.9]

input = wlrec[0]
pred = neural_network(input, weights)

print(pred)
````

### Hasil

Output:

```text
[0.195, 0.13, 0.585]
```

Perhitungannya:

- `0.65 × 0.3 = 0.195`
- `0.65 × 0.2 = 0.13`
- `0.65 × 0.9 = 0.585`

Satu input dapat menghasilkan beberapa prediction karena setiap output mempunyai weight yang berbeda.

## 5. Multiple Inputs dan Multiple Outputs

Ketika terdapat banyak input dan banyak output, weight direpresentasikan sebagai **matrix**.

Contohnya:

```python
weights = [
    [0.1, 0.1, -0.3],
    [0.1, 0.2,  0.0],
    [0.0, 1.3,  0.1]
]
```

Setiap baris berisi weight yang digunakan untuk menghasilkan satu output.

Operasi ini menjadi dasar dari banyak perhitungan dalam neural network.

````python
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output

def vect_mat_mul(vect, matrix):
    assert(len(vect) == len(matrix))
    output = [0, 0, 0]

    for i in range(len(vect)):
        output[i] = w_sum(vect, matrix[i])

    return output

def neural_network(input, weights):
    pred = vect_mat_mul(input, weights)
    return pred

weights = [
    [0.1, 0.1, -0.3],
    [0.1, 0.2,  0.0],
    [0.0, 1.3,  0.1]
]

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0], wlrec[0], nfans[0]]
pred = neural_network(input, weights)

print(pred)
````

### Hasil

Output:

```text
[0.555, 0.98, 0.965]
```

Ketiga nilai tersebut merupakan prediction untuk tiga output. Setiap output dihitung menggunakan weighted sum antara input dan baris weight yang sesuai.

## 6. Predicting on Predictions

Neural network dapat disusun menjadi beberapa tahap. Output dari satu tahap digunakan sebagai input untuk tahap berikutnya.

Strukturnya:

```text
Input
  ↓
Layer 1
  ↓
Hidden representation
  ↓
Layer 2
  ↓
Prediction
```

Konsep ini disebut **predicting on predictions**. Dengan menyusun beberapa operasi, network dapat membangun representasi yang lebih kompleks daripada network satu layer.

````python
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output

def vect_mat_mul(vect, matrix):
    assert(len(vect) == len(matrix))
    output = [0, 0, 0]

    for i in range(len(vect)):
        output[i] = w_sum(vect, matrix[i])

    return output

def neural_network(input, weights):
    hid = vect_mat_mul(input, weights[0])
    pred = vect_mat_mul(hid, weights[1])
    return pred

ih_wgt = [
    [0.1, 0.2, -0.1],
    [-0.1, 0.1, 0.9],
    [0.1, 0.4, 0.1]
]

hp_wgt = [
    [0.3, 1.1, -0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
]

weights = [ih_wgt, hp_wgt]

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0], wlrec[0], nfans[0]]
pred = neural_network(input, weights)

print(pred)
````

### Hasil

Output:

```text
[0.2135, 0.14500000000000002, 0.506]
```

Prosesnya berlangsung dalam dua tahap:

$$hidden = input \times weights_{input-hidden}$$

kemudian:

$$prediction = hidden \times weights_{hidden-output}$$

Jadi prediction akhir berasal dari hasil prediction sebelumnya.

## 7. Forward Propagation

Proses ketika input bergerak maju melalui layer-layer neural network sampai menghasilkan output disebut **forward propagation**.

Gambaran sederhananya:

```text
Input
  ↓
Weighted Sum
  ↓
Hidden Layer
  ↓
Weighted Sum
  ↓
Prediction
```



## 8. Intuisi Weight

Weight menentukan seberapa besar kontribusi sebuah input terhadap prediction.

- Weight positif → input positif cenderung menaikkan prediction.
- Weight negatif → input positif cenderung menurunkan prediction.
- Weight mendekati nol → pengaruh input relatif kecil.

Namun pengaruh sebenarnya berasal dari kombinasi:

$$input \times weight$$

Karena itu, prediction ditentukan oleh **nilai input sekaligus weight**.

## 9. Teori yang Perlu Dipahami

### Weighted Sum
Gabungan beberapa input yang masing-masing dikalikan dengan weight.

$$z = \sum_i x_iw_i$$

### Dot Product
Operasi matematika yang melakukan perkalian elemen-per-elemen lalu menjumlahkannya.

$$x \cdot w = \sum_i x_iw_i$$

### Vector
Sekumpulan angka yang dapat merepresentasikan input atau prediction.

### Matrix
Kumpulan vector yang digunakan untuk merepresentasikan banyak weight.

### Vector-Matrix Multiplication
Operasi yang mengubah satu vector input menjadi vector output menggunakan matrix weight.

### Hidden Layer
Bagian di antara input dan output yang menghasilkan representasi intermediate.

### Forward Propagation
Proses meneruskan input melalui layer secara berurutan hingga menghasilkan prediction.
