## Fundamental Concepts: How Do Machines Learn?

### Tujuan
Chapter ini membangun kosakata dasar untuk memahami **artificial intelligence (AI)**, **machine learning (ML)**, dan **deep learning (DL)**. Fokus utamanya adalah memahami dua cara mengelompokkan algoritma machine learning:

1. **Supervised vs. unsupervised learning**
2. **Parametric vs. nonparametric learning**

Chapter ini juga memperkenalkan pola **predict → compare → learn** yang akan menjadi fondasi untuk pembahasan neural network pada chapter berikutnya.

## 1. Apa Itu Deep Learning?

**Deep learning adalah subset dari machine learning.**

Deep learning berada di dalam bidang machine learning dan terutama menggunakan **artificial neural networks**, yaitu kelas algoritma yang secara longgar terinspirasi dari cara kerja otak manusia.

Hubungan konsepnya dapat dipahami sebagai:

**Artificial Intelligence → Machine Learning → Deep Learning**

Tidak semua artificial intelligence merupakan deep learning, dan tidak semua penggunaan deep learning bertujuan membangun kecerdasan umum. Deep learning dapat digunakan untuk tugas-tugas praktis seperti:

- **Computer vision** → data berupa gambar
- **Natural language processing** → data berupa teks
- **Automatic speech recognition** → data berupa audio


## 2. Apa Itu Machine Learning?

Machine learning berfokus pada membuat mesin mampu melakukan tugas yang **tidak diprogram secara eksplisit untuk setiap kasusnya**.

Gagasan sederhananya adalah:

> mesin mengamati pola pada data, kemudian berusaha meniru atau menggunakan pola tersebut.

Terdapat dua bentuk utama machine learning:

| Jenis | Gagasan utama |
|---|---|
| **Supervised learning** | Mempelajari hubungan antara data input dan output yang diketahui |
| **Unsupervised learning** | Mencari struktur atau kelompok dalam data tanpa output benar yang telah ditentukan |

Perbedaan pentingnya bukan sekadar apakah data memiliki angka atau label, tetapi **jenis pola yang ingin dipelajari oleh algoritma**.

## 3. Supervised Machine Learning

Pada **supervised learning**, model belajar mengubah satu dataset menjadi dataset lain yang menjadi target.

Secara konseptual:

**Input data → Machine Learning → Output / Prediction**

Contohnya, kita dapat memiliki data yang dapat diamati lalu ingin menghasilkan informasi yang tidak langsung tersedia dari data tersebut.

Hal terpenting adalah terdapat **pola atau target yang ingin ditiru**. Model membuat prediksi, kemudian prediksi tersebut dapat dibandingkan dengan kondisi sebenarnya.

### Inti pemahaman

Supervised learning = **belajar memprediksi satu dataset berdasarkan dataset lainnya**.

## 4. Unsupervised Machine Learning

Pada **unsupervised learning**, model tidak diberikan jawaban benar yang sudah ditentukan.

Tujuannya adalah:

> **mencari struktur atau pola di dalam data.**

Salah satu bentuk paling mudah untuk membayangkannya adalah **clustering**.

Misalnya terdapat data:

- puppies
- pizza
- kittens
- hot dog
- burger

Algoritma dapat mengelompokkan data tersebut berdasarkan pola yang ditemukan. Hasilnya dapat berupa label cluster seperti `1` dan `2`, tetapi algoritma tidak harus mengetahui nama manusia untuk cluster tersebut.

Contoh interpretasi hasil:

- Cluster 1 → sesuatu yang cute
- Cluster 2 → sesuatu yang delicious

Yang penting, label tersebut berasal dari struktur yang ditemukan pada data, bukan dari jawaban benar yang diberikan sebelumnya.


## 5. Supervised vs. Unsupervised

Kedua jenis pembelajaran tersebut dapat dibandingkan seperti berikut:

| Aspek | Supervised | Unsupervised |
|---|---|---|
| Target benar tersedia? | Ya | Tidak |
| Tujuan utama | Memprediksi output | Menemukan struktur |
| Bentuk sederhana | Input → target | Data → cluster/struktur |
| Contoh konsep | Prediction | Clustering |
| Pertanyaan | “Apa output yang seharusnya?” | “Pola apa yang ada di data?” |


## 6. Parametric vs. Nonparametric Learning

Selain supervised dan unsupervised, algoritma machine learning juga dapat dibedakan berdasarkan bagaimana pembelajarannya disimpan.

### Parametric

Model **parametric** memiliki jumlah parameter yang tetap.

Parameter dapat dibayangkan sebagai **knob** yang nilainya dapat diubah untuk membuat prediksi menjadi lebih baik.

Gambaran sederhananya:

**Data → Model dengan sejumlah knob → Prediction**

Pembelajaran dilakukan dengan mengubah nilai knob tersebut.

### Nonparametric

Pada model **nonparametric**, jumlah parameter bergantung pada data.

Pendekatan ini sering dapat dibayangkan sebagai proses **counting**. Ketika struktur atau kategori baru muncul pada data, jumlah informasi/parameter yang diperlukan dapat bertambah.

Jadi, perbedaan utamanya:

| Parametric | Nonparametric |
|---|---|
| Jumlah parameter tetap | Jumlah parameter ditentukan/dipengaruhi data |
| Dapat dibayangkan sebagai “turning knobs” | Dapat dibayangkan sebagai “counting” |
| Deep learning termasuk model parametric | Tidak menjadi fokus lanjutan buku ini |

## 7. Empat Kombinasi Jenis Machine Learning

Karena supervised/unsupervised dan parametric/nonparametric merupakan dua cara pengelompokan yang berbeda, keduanya dapat dikombinasikan:

| | Parametric | Nonparametric |
|---|---|---|
| **Supervised** | Supervised parametric | Supervised nonparametric |
| **Unsupervised** | Unsupervised parametric | Unsupervised nonparametric |

Jadi, sebuah algoritma dapat dijelaskan berdasarkan **apa yang dipelajari** dan **bagaimana pembelajarannya direpresentasikan**.

- **Supervised vs. unsupervised** → jenis pola yang dipelajari
- **Parametric vs. nonparametric** → bagaimana pembelajaran disimpan dan bagaimana jumlah parameter ditentukan

## 8. Supervised Parametric Learning

Supervised parametric learning menggunakan model dengan jumlah knob yang tetap.

Contoh konseptualnya adalah memprediksi apakah sebuah tim olahraga akan menang.

Model menerima berbagai data, misalnya:

- lokasi pertandingan
- lawan
- statistik pemain
- jumlah penggemar

Model kemudian menghasilkan prediksi, misalnya **98% peluang menang**.

Pembelajaran berlangsung melalui tiga langkah penting:

### Step 1 — Predict
Model menggunakan data input untuk menghasilkan prediksi.

### Step 2 — Compare to the truth pattern
Prediksi dibandingkan dengan kondisi sebenarnya.

Misalnya:

**Prediction = 98%**

tetapi hasil sebenarnya adalah:

**Truth = 0%**

Artinya prediksi meleset cukup jauh.

### Step 3 — Learn
Model mengubah nilai parameter/knob berdasarkan kesalahan tersebut agar prediksi berikutnya menjadi lebih baik.

Inilah pola:

**Predict → Compare → Learn**

Pola ini sangat penting karena akan terus digunakan ketika mulai membangun neural network.

## 9. Unsupervised Parametric Learning

Unsupervised parametric learning juga menggunakan parameter/knob, tetapi tujuannya bukan mencocokkan prediksi dengan target yang sudah diketahui.

Tujuannya adalah **mengelompokkan data**.

Misalnya model ingin menemukan tiga kelompok dari data pertandingan. Untuk setiap kelompok, model dapat menghasilkan nilai yang menggambarkan seberapa kuat sebuah data termasuk ke kelompok tersebut.

Contoh:

| Group | Probability |
|---|---:|
| Group 1 | 94% |
| Group 2 | 1% |
| Group 3 | 5% |

Data tersebut paling kuat diasosiasikan dengan Group 1.

Intinya:

**Unsupervised + Parametric = menggunakan parameter untuk mempelajari pengelompokan data.**

## 10. Nonparametric Learning

Nonparametric learning menggunakan jumlah parameter yang bergantung pada data.

Contoh sederhana yang digunakan untuk memahami ide ini adalah **menghitung hubungan antara warna lampu lalu lintas dan apakah mobil berjalan atau berhenti**.

Misalnya model mencatat berapa kali setiap warna lampu menyebabkan mobil “go”.

Jika terdapat tiga warna/kategori, dapat terdapat tiga nilai hitungan. Jika data memiliki lima kategori, jumlah nilai yang diperlukan dapat menjadi lima.

Dengan demikian, jumlah parameter berubah mengikuti struktur data.

### Hal penting tentang istilah parameter

Parameter adalah sekumpulan angka yang digunakan untuk merepresentasikan pola.

Parameter dapat berupa:

- weights
- counts
- normalized counts
- correlation coefficients

Deep learning termasuk **parametric models**.

## 11. Cara Cepat Membedakan Keempat Konsep

Gunakan dua pertanyaan:

### Pertanyaan 1 — Apakah target/output yang benar sudah diketahui?

- **Ya** → Supervised
- **Tidak** → Unsupervised

### Pertanyaan 2 — Apakah jumlah parameter ditetapkan secara tetap?

- **Ya** → Parametric
- **Tidak, mengikuti data** → Nonparametric

Maka:

| Kondisi | Jenis |
|---|---|
| Target diketahui + parameter tetap | Supervised Parametric |
| Target diketahui + parameter mengikuti data | Supervised Nonparametric |
| Target tidak diketahui + parameter tetap | Unsupervised Parametric |
| Target tidak diketahui + parameter mengikuti data | Unsupervised Nonparametric |

## 12. Konsep Terpenting: Predict → Compare → Learn

Konsep ini menjadi jembatan dari pembahasan konseptual menuju neural network.

### Predict
Model menggunakan input untuk menghasilkan output.

### Compare
Output dibandingkan dengan pola/target yang diinginkan.

### Learn
Model mengubah parameter untuk mengurangi kesalahan dan menghasilkan prediksi yang lebih baik.
