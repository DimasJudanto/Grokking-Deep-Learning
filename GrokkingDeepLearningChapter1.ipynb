## 1. Why You Should Learn Deep Learning

Deep learning dipandang sebagai alat untuk melakukan **incremental automation of intelligence**, yaitu mengotomatisasi kemampuan tertentu secara bertahap melalui pembelajaran dari data.

Perkembangannya telah menghasilkan kemajuan pada berbagai bidang, seperti:

- **Computer vision** untuk memahami informasi dari gambar.
- **Speech recognition** untuk memahami atau mengolah suara.
- **Machine translation** untuk menerjemahkan bahasa secara otomatis.

Salah satu karakteristik pentingnya adalah penggunaan neural network pada berbagai jenis permasalahan. Karena itu, deep learning dapat digunakan pada banyak domain selama data memiliki pola yang dapat dipelajari.

Deep learning juga memiliki potensi untuk mengotomatisasi sebagian pekerjaan yang sebelumnya dilakukan manusia. Hal ini membuat pemahaman terhadap teknologi deep learning menjadi semakin relevan dalam perkembangan industri.

### Deep Learning sebagai Bidang yang Kreatif

Deep learning tidak hanya berkaitan dengan otomatisasi. Teknologi ini juga dapat digunakan untuk membangun sistem yang meniru atau menghasilkan pola yang sebelumnya dibuat manusia.

Hal tersebut membuat deep learning menjadi bidang yang mempertemukan **manusia, data, dan komputer**, karena kemampuan model diperoleh melalui proses pembelajaran dari data dan eksperimen.

## 2. Will This Be Difficult to Learn?

Pembelajaran deep learning tidak harus dimulai dari matematika tingkat lanjut. Materi dibangun secara bertahap melalui konsep sederhana, contoh, dan implementasi.

Proses belajar diarahkan agar pembaca dapat melihat secara langsung bagaimana model yang dibuat dapat belajar. Setelah memahami konsep awal, pembahasan bergerak menuju neural network, prediksi, evaluasi kesalahan, dan proses learning.

Dengan demikian, pemahaman deep learning berkembang dari konsep dasar menuju implementasi yang semakin kompleks.

## 3. What You Need to Get Started

Beberapa persiapan dasar yang diperlukan adalah:

### Jupyter Notebook
Digunakan sebagai lingkungan interaktif untuk menjalankan dan mengamati kode serta melihat bagian-bagian model ketika bekerja.

### NumPy
Digunakan sebagai library numerik untuk operasi data dan matriks. Penggunaannya membantu memperlihatkan proses deep learning tanpa langsung menyembunyikan mekanismenya di balik framework tingkat tinggi.

### Dasar Matematika
Materi mengasumsikan pemahaman **basic algebra** atau matematika tingkat sekolah menengah.

### Python
Contoh implementasi menggunakan Python, sehingga kemampuan membaca dan menulis kode Python dasar diperlukan.

### Masalah yang Ingin Diselesaikan
Memiliki masalah atau proyek yang menarik dapat menjadi konteks untuk memahami bagaimana deep learning digunakan untuk mempelajari pola dari data.

## 4. Python sebagai Dasar Implementasi

Python digunakan sebagai bahasa utama dalam contoh karena sintaksnya relatif mudah dibaca dan banyak digunakan dalam machine learning.

Kemampuan Python yang dibutuhkan pada tahap awal tidak harus mendalam. Yang penting adalah mampu membaca struktur kode dasar sehingga perhatian dapat difokuskan pada konsep deep learning.

Chapter selanjutnya mulai membawa konsep tersebut ke pembahasan mengenai bagaimana mesin belajar.
